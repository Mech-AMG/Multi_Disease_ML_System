```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# Locate the project root
current_folder = Path.cwd()

project_root = (
    current_folder.parent
    if current_folder.name == "notebooks"
    else current_folder
)

# Build the path to the raw heart disease dataset
data_path = (
    project_root
    / "data"
    / "raw"
    / "heart_disease_uci.csv"
)

# Verify the dataset path
print(
    "Dataset path:",
    data_path
)

print(
    "File exists:",
    data_path.exists()
)

# Load the raw dataset
heart_df = pd.read_csv(
    data_path
)

# Display the basic dataset information
print(
    "\nDataset shape:",
    heart_df.shape
)

print(
    "\nColumn names:"
)

print(
    heart_df.columns.tolist()
)

print(
    "\nData types:"
)

print(
    heart_df.dtypes
)

print(
    "\nMissing values by column:"
)

print(
    heart_df.isnull()
    .sum()
    .sort_values(
        ascending=False
    )
)

print(
    "\nDuplicated rows:",
    heart_df.duplicated().sum()
)

print(
    "\nTarget distribution:"
)

print(
    heart_df["num"]
    .value_counts(
        dropna=False
    )
    .sort_index()
)

print(
    "\nRecords by dataset source:"
)

print(
    heart_df["dataset"]
    .value_counts(
        dropna=False
    )
)

display(
    heart_df.head()
)
```

```python
# Create a binary heart disease target
heart_binary_target = (
    heart_df["num"] > 0
).astype(int)

# Display the binary target distribution
binary_target_summary = pd.DataFrame(
    {
        "Count": (
            heart_binary_target
            .value_counts()
            .sort_index()
        ),
        "Percentage": (
            heart_binary_target
            .value_counts(
                normalize=True
            )
            .sort_index()
            .mul(100)
            .round(2)
        )
    }
)

binary_target_summary.index = [
    "No heart disease",
    "Heart disease present"
]

print(
    "Binary target distribution:"
)

display(
    binary_target_summary
)

# Display target distribution by dataset source
source_target_table = pd.crosstab(
    heart_df["dataset"],
    heart_binary_target,
    margins=True
)

source_target_table.columns = [
    "No heart disease",
    "Heart disease present",
    "Total"
]

print(
    "Target distribution by dataset source:"
)

display(
    source_target_table
)

# Inspect all categorical feature values
categorical_columns = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "thal"
]

print(
    "Categorical feature values:"
)

for column in categorical_columns:
    print(
        f"\n{column}:"
    )

    print(
        heart_df[column]
        .value_counts(
            dropna=False
        )
    )

# Inspect numerical ranges and suspicious zero values
numerical_columns = [
    "age",
    "trestbps",
    "chol",
    "thalch",
    "oldpeak",
    "ca"
]

numerical_summary = pd.DataFrame(
    {
        "Minimum": (
            heart_df[
                numerical_columns
            ].min()
        ),
        "Maximum": (
            heart_df[
                numerical_columns
            ].max()
        ),
        "Mean": (
            heart_df[
                numerical_columns
            ].mean()
        ),
        "Missing Values": (
            heart_df[
                numerical_columns
            ].isnull().sum()
        ),
        "Zero Values": (
            heart_df[
                numerical_columns
            ].eq(0).sum()
        )
    }
)

print(
    "\nNumerical feature summary:"
)

display(
    numerical_summary.round(2)
)
```

```python
# Create a working copy of the raw dataset
heart_clean = heart_df.copy()

# Create the binary target
heart_clean["heart_disease"] = (
    heart_clean["num"] > 0
).astype("int8")

# Remove the identifier and original multiclass target
heart_clean = heart_clean.drop(
    columns=[
        "id",
        "num"
    ]
)

# Replace medically impossible zero values with missing values
heart_clean["trestbps"] = (
    heart_clean["trestbps"]
    .replace(
        0,
        np.nan
    )
)

heart_clean["chol"] = (
    heart_clean["chol"]
    .replace(
        0,
        np.nan
    )
)

# Convert nullable boolean features to readable categories
boolean_mapping = {
    True: "Yes",
    False: "No"
}

for column in [
    "fbs",
    "exang"
]:
    heart_clean[column] = (
        heart_clean[column]
        .map(
            boolean_mapping
        )
    )

# Arrange the dataset columns
heart_clean = heart_clean[
    [
        "dataset",
        "age",
        "sex",
        "cp",
        "trestbps",
        "chol",
        "fbs",
        "restecg",
        "thalch",
        "exang",
        "oldpeak",
        "slope",
        "ca",
        "thal",
        "heart_disease"
    ]
]

# Build the processed dataset path
processed_data_path = (
    project_root
    / "data"
    / "processed"
    / "heart_disease_clean.csv"
)

# Save the cleaned dataset without imputing missing values
heart_clean.to_csv(
    processed_data_path,
    index=False
)

# Verify the cleaned dataset
print(
    "Cleaned dataset shape:",
    heart_clean.shape
)

print(
    "\nMissing values after cleaning:"
)

print(
    heart_clean.isnull()
    .sum()
    .sort_values(
        ascending=False
    )
)

print(
    "\nTarget distribution:"
)

print(
    heart_clean[
        "heart_disease"
    ].value_counts()
    .sort_index()
)

print(
    "\nProcessed file saved:",
    processed_data_path.exists()
)

print(
    "Processed file path:",
    processed_data_path
)

display(
    heart_clean.head()
)
```
